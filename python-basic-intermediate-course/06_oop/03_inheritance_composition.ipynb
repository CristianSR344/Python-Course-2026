# Inheritance and Composition
## Objectives
Specialize behavior with inheritance and combine objects with composition/polymorphism.

```python
class Notification:
    def send(self, message): raise NotImplementedError
class ConsoleNotification(Notification):
    def send(self, message): return f"CONSOLE: {message}"
class TicketService:
    def __init__(self, notifier): self.notifier = notifier
    def close(self, title): return self.notifier.send(f"Closed: {title}")
print(TicketService(ConsoleNotification()).close("Login issue"))
```
