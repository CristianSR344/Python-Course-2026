# Classes and Objects
## Objectives
Define classes, initialize objects, use attributes/methods, class attributes, and `__str__`.

```python
class User:
    category = "portal-user"
    def __init__(self, name, email):
        self.name, self.email = name, email
    def __str__(self):
        return f"{self.name} <{self.email}>"

user = User("Ada", "ada@example.com")
print(user, user.category)
```
