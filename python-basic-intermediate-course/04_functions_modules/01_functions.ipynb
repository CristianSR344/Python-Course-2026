## Objectives

```python
def full_name(first, last, uppercase=False):
    name = f"{first.strip()} {last.strip()}"
    return name.upper() if uppercase else name.title()

print(full_name("ada", "lovelace"))
print(full_name(last="hopper", first="grace", uppercase=True))
```
