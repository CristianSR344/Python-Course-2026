# CSV and JSON
## Objectives
Read tabular CSV, serialize JSON, and understand API-shaped data.

```python
import csv, io, json
csv_text = "id,name\n1,Ana\n2,Ben\n"
users = list(csv.DictReader(io.StringIO(csv_text)))
json_text = json.dumps(users, indent=2)
print(json_text)
print(json.loads(json_text)[0]["name"])
```
