# Exceptions
## Objectives
Use `try`, specific `except`, `else`, and `finally` to handle expected errors.

```python
raw_age = "twenty"
try:
    age = int(raw_age)
except ValueError as error:
    print(f"Invalid age: {error}")
else:
    print(age)
finally:
    print("Validation complete")
```
