# Variables and Types
## Objectives
Create variables; inspect strings, integers, floats, booleans, and `None`; convert types; format output.

```python
employee_name = "Jordan Lee"
age = 28
hourly_rate = 24.5
active = True
manager = None
for value in (employee_name, age, hourly_rate, active, manager):
    print(value, type(value).__name__)
```

A variable is a readable label for a value. Python determines the type at runtime. Use `int()` or `float()` to convert numeric text.

```python
years_text = "4"
years = int(years_text)
print(f"Next year: {years + 1}")
```
