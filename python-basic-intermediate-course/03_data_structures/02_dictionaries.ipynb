# Dictionaries and Nested Data
## Objectives
Create, update, search, and iterate through dictionaries and lists of dictionaries.

```python
user = {"id": 1, "name": "Ada", "department": "IT"}
user["active"] = True
for key, value in user.items():
    print(key, value)
users = [user, {"id": 2, "name": "Lin", "department": "HR", "active": False}]
print([item["name"] for item in users if item["active"]])
```
